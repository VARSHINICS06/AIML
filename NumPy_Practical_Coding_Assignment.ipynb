Question 1: Array Creation, Reshaping, and Inspection
Create a 1D NumPy array containing all multiples of 5 starting from 5 up to Reshape this 1D array into a 2D matrix with 100 (inclusive).
4 rows and 5 columns.

Write code to display:

The reshaped 4 × 5 matrix




The number of dimensions ( ndim )

The shape of the matrix ( shape )

The total number of elements ( size )

The data type of the elements ( dtype )

```python
import numpy as np
arr=np.arange(5,101,5)
print(arr)
b=arr.reshape(4,5)
print("4 x 5 Matrix")
print(b)
print("Shape",b.shape)
print("Size",b.size)
print("Dimension",b.ndim)
print("dtype",b.dtype)
```

 Replace the inner 3 × 3 center block with 0s.

**Question 3: Element-wise Arithmetic and Aggregations**

You are given two 1D NumPy arrays representing the unit prices (in ₹) and quantities sold for 5 products:

prices = np.array([120, 250, 80, 450, 300])
 quantities = np.array([4, 2, 10, 1, 3])



An array representing the total revenue generated by each individual product.

The overall combined revenue across all 5 products.

 The average revenue per product. The index position of the highest-earning product using

```python
prices = np.array([120, 250, 80, 450, 300])
quantities = np.array([4, 2, 10, 1, 3])
revenue=prices*quantities
print(revenue)
total_revenue=np.sum(revenue)
avg_revenue=np.mean(revenue)
highest_index=np.argmax(revenue)
print("Individual",revenue)
print("Total Revenue",total_revenue)
print("Average Revenue",avg_revenue)
print("Highest Earning Product",highest_index)
```

**Question 4: Boolean Masking and Data Cleaning**

Given:
rainfall = np.array([
    12.5, -1.0, 0.0, 18.2, -99.0,
    5.4, 22.1, 0.0, -1.0, 14.8
])

Requirements:
1. Extract valid rainfall readings.
2. Replace negative values with 0.
3. Count days with rainfall greater than 10 mm


```python
rainfall = np.array([
    12.5, -1.0, 0.0, 18.2, -99.0,
    5.4, 22.1, 0.0, -1.0, 14.8
])
valid=rainfall[rainfall>=0]
print("Valid Rainfall:", valid)
# Replace negative values with 0
rainfall[rainfall<0]=0.0

print("Cleaned Rainfall:", rainfall)

heavy=np.sum(rainfall>10)
print("Days with heavy rainfall:", heavy)
```

**Question 5: Conditional Array Operations with np.where()**

Generate 12 random integers between 20 and 100

Print the original array of scores.

Create and print a new array named status that assigns to 40, and "Fail" to scores below 40 .

Create and print a new array named "Pass" to scores greater than or equal updated_scores where any student who scored below receives 5 grace marks, while scores of 40 and above remain unchanged.


```python

import numpy as np

np.random.seed(10)

scores = np.random.randint(20, 101, 12)

status = np.where(scores >= 40, "Pass", "Fail")

updated_scores = np.where(scores < 40, scores + 5, scores)

print("Original Scores:", scores)
print("Status:", status)
print("Updated Scores:", updated_scores)
```

**Question 6: Row-wise and Column-wise Aggregations**

Generate a 6 × 4 matrix containing random marks between 50 and 100, inclusive.



```python
marks = np.random.randint(50, 101, (6, 4))

print("Marks Matrix:")
print(marks)

# Average marks of each student
student_avg = np.round(np.mean(marks, axis=1), 2)

print("Student Average:")
print(student_avg)

# Highest mark in each subject
subject_max = np.max(marks, axis=0)

print("Highest Marks in Each Subject:")
print(subject_max)

# Students whose average is greater than 75
selected_students = marks[student_avg > 75]

print("Students with Average Above 75:")
print(selected_students)
```

**Question 7: Feature Scaling Using Broadcasting**

```python
houses = np.array([
    [1200, 2, 55],
    [1800, 3, 85],
    [950,  1, 40],
    [2400, 4, 120],
    [1500, 3, 70]
])
X_min=np.min(houses,axis=0)

X_max=np.max(houses,axis=0)


scaled=(houses-X_min)/(X_max-X_min)

print("Minimum:", X_min)
print("Maximum:", X_max)
print("Scaled Data:\n", np.round(scaled, 3))
```

**Question 8: Combining and Stacking Arrays**

Requirements:
1. Combine morning and evening sales horizontally.
2. Add the fourth branch.
3. Calculate total sales for each column.

```python
morning_sales = np.array([
    [15, 20],
    [10, 12],
    [25, 30]
])

evening_sales = np.array([
    [18, 22],
    [14, 15],
    [20, 28]
])
full_day=np.concatenate((morning_sales,evening_sales),axis=1)
print("Full Day Sales:")
print(full_day)
final_sale=np.vstack((full_day,[12,16,15,19]))
print("Final Sales Matrix:")
print(final_sale)

# Total sales column-wise
total = np.sum(final_sale, axis=0)

print("Total Sales in Each Column:")
print(total)
```

**Question 9: Unique Elements and Frequency Distribution**

```python
ratings = np.array([
    5, 4, 3, 5, 2, 5, 4, 1, 3, 5,
    4, 4, 2, 5, 3, 4, 5, 1, 4, 3
])
unique,counts=np.unique(ratings,return_counts=True)
print("Unique Ratings:", unique)
print("Frequency Distribution:", counts)
percentage=(counts/ratings.size)*100
print("Percentage:",percentage)
# Display summary
summary = np.column_stack(
    (unique, counts, percentage)
)

print("Rating | Count | Percentage")
print(summary)
```

**Question 10: Indirect Sorting and Top-K Selection**

Requirements:
1. Sort scores from highest to lowest.
2. Reorder employee IDs accordingly.
3. Display the top three employees.

```python
emp_ids = np.array([
    "E101", "E102", "E103",
    "E104", "E105", "E106"
])

scores = np.array([78, 92, 65, 88, 95, 71])

indices=np.argsort(scores)[::-1]
print(indices)
sorted_ids=emp_ids[indices]
sorted_scores=scores[indices]
print("Sorted Employee IDs:", sorted_ids)
print("Sorted Scores:", sorted_scores)
# Top 3 employees
top_ids = sorted_ids[:3]
top_scores = sorted_scores[:3]

print("Top 3 Employees:")
print(top_ids)
print(top_scores)
```
